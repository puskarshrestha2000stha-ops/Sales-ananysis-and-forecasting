sales calculation

```python
sales_columns = [
    'S-P1',
    'S-P2',
    'S-P3',
    'S-P4'
]


quantity_columns = [
    'Q-P1',
    'Q-P2',
    'Q-P3',
    'Q-P4'
]


df['Total_Sales'] = df[sales_columns].sum(axis=1)


df['Total_Quantity'] = df[quantity_columns].sum(axis=1)
```

```python
df.head()
```

product performance

```python
product = pd.DataFrame({

    'Product':
    ['P1','P2','P3','P4'],

    'Total Sales':
    [
        df['S-P1'].sum(),
        df['S-P2'].sum(),
        df['S-P3'].sum(),
        df['S-P4'].sum()
    ],

    'Total Quantity':
    [
        df['Q-P1'].sum(),
        df['Q-P2'].sum(),
        df['Q-P3'].sum(),
        df['Q-P4'].sum()
    ]

})
product
```

product sales graph


```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=product,
    x='Product',
    y='Total Sales'
)

plt.title("Product Wise Sales")

plt.show()
```

monthly sales trend

```python
df['Date'] = pd.to_datetime(df['Date'], errors='ignore')
```

```python
monthly_sales = (
    df.groupby(
        df['Date'].dt.to_period('M')
    )
    ['Total_Sales']
    .sum()
)
plt.figure(figsize=(12,5))
monthly_sales.plot(
    marker='o'
)

plt.title(
    "Monthly Sales Trend"
)

plt.xlabel("Month")

plt.ylabel("Sales")

plt.xticks(rotation=45)

plt.show()
```

Quantity analysis

```python
plt.figure(figsize=(8,5))
sns.barplot(
    data=product,
    x='Product',
    y='Total Quantity'
)
plt.title(
    "Product Quantity Sold"
)
plt.show()
```

Corelation analysis

```python
plt.figure(figsize=(10,6))


sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap='coolwarm')
plt.title(
    "Correlation Analysis")
plt.show()
```

Outlier check

```python
plt.figure(figsize=(8,5))
sns.boxplot(
    data=df[['Total_Sales','Total_Quantity']])
plt.title("Outlier Detection")


plt.show()
```

Final business insight

```python
best_product = product.loc[
    product['Total Sales'].idxmax()
]


final_report = pd.DataFrame({

    'Metric':
    [
        'Total Revenue',
        'Total Quantity Sold',
        'Average Sales',
        'Best Selling Product'
    ],


    'Value':
    [
        df['Total_Sales'].sum(),

        df['Total_Quantity'].sum(),

        df['Total_Sales'].mean(),

        best_product['Product']
    ]

})
final_report
```
